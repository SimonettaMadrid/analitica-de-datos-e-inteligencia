```python
### Cargar paqueterias 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
###cargar datos 
data=pd.read_csv('cuentas_credicel.csv',encoding='latin-1')
```

```python
data.head()
```

```python
###una copia de los datos
data2=data.copy()
```

```python
###aplicar filtros 
##filtro 1 todo menos folio folio folio_solicitud, unamed33
filtro1 = data.drop(columns=['folio', 'folio_solicitud', 'Unnamed: 33',  'Unnamed: 34', 'Unnamed: 35', 'Unnamed: 36'])
filtro1.head()
```

```python
###todas las filas multiplo de 4 
filtro2 = filtro1.iloc[::4, :]
filtro2.head()
```

```python
###registros con la marca igual a ALCATEL, MOTOROLA, NOKIA, ZTE,SAMSUNG
filtro3 = filtro2[filtro2['marca'].isin(['ALCATEL', 'MOTOROLA', 'NOKIA', 'ZTE','SAMSUNG'])]
filtro3.head()
```

```python
### registros entre 500 y 2000
filtro4 = filtro3[(filtro3['enganche'] >= 500) & (filtro3['enganche'] <= 2000)]
filtro4.head()
```

```python
### todos los registros en la categoria en riesgo no sea cancelado o fraude 
filtro5 = filtro4[~filtro4['riesgo'].isin(['Cancelado', 'Fraude'])]
```

```python
pd.set_option("display.max_rows", None)
filtro5['riesgo'].value_counts()
```

```python
filtro5['fecha'].value_counts()
```

```python
###todos los registros cuya fecha sea posterior al 15/11/2021
filtro6 = filtro5[filtro5['fecha'] > '2021-11-15']
```
