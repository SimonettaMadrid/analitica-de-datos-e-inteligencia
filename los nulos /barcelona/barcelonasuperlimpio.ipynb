```python
#Importamos las librerias pandas, numpy y matplotlib respectivamente
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
###cargamos los datos
data = pd.read_csv('listingsbarcelona.csv')
```

```python
pd.set_option("display.max_rows", None)
data.isnull().sum().sort_values(ascending=False)
```

```python
data1 = data.copy()
```

#### para neighbourhood Barcelona, Catalonia, Spain
#### para host_neighbourhood se elaboro el barrio mas comun y se inserto como dato random para mejorar la credibilidad y eliminar los nulos 

```python
data1["neighbourhood"] =data1["neighbourhood"].fillna("Barcelona, Catalonia, Spain")
data1["neighbourhood"].isnull().sum() 
```

```python
barrio_mas_comun = data1['host_neighbourhood'].mode()[0]
data1['host_neighbourhood'] = data1['host_neighbourhood'].fillna(barrio_mas_comun)
data1['host_neighbourhood'].isnull().sum()
```

### para estas dos variables se utiliza la palabra no dato ya qye si los eiminamos nos podriamos llevar parte importante de ellos neighborhood_overview ,host_about 

```python
data1["host_about"] =data1["host_about"].fillna("No Data")
data1["host_about"].isnull().sum()
```

```python
                        
data1["neighborhood_overview"] =data1["neighborhood_overview"].fillna("No Data")
data1["neighborhood_overview"].isnull().sum() 

```

#### license con exempt que en el diccionario significa no registro

```python
data1["license"] = data1["license"].fillna("exempt")
data1["license"].isnull().sum()
```

### las numericas 
Para las numericas utilice el mejor metodo etre ellas para imputar datos
### review_scores_value, review_scores_accuracy,review_scores_checkin
### review_scores_communication, review_scores_cleanliness
### review_scores_location review_scores_rating                      reviews_per_month, review_scores_location

```python
data1['review_scores_rating'] = data1['review_scores_rating'].fillna(data1['review_scores_rating'].mean())
data1['review_scores_rating'].isnull().sum()
```

```python
data1['review_scores_communication'] = data1['review_scores_communication'].fillna(data1['review_scores_communication'].mean())
data1['review_scores_communication'].isnull().sum()
```

```python
data1['review_scores_cleanliness'] = data1['review_scores_cleanliness'].fillna(data1['review_scores_cleanliness'].mean())
data1['review_scores_cleanliness'].isnull().sum()
```

```python
data1['review_scores_value'] = data1['review_scores_value'].fillna(data1['review_scores_value'].mean())
data1['review_scores_value'].isnull().sum()
```

```python
data1['review_scores_accuracy'] = data1['review_scores_accuracy'].fillna(data1['review_scores_accuracy'].mean())
data1['review_scores_accuracy'].isnull().sum()
```

```python
data1['review_scores_checkin'] = data1['review_scores_checkin'].fillna(data1['review_scores_checkin'].mean())
data1['review_scores_checkin'].isnull().sum()
```

### eliminar los porcentajes para eliminar nulos y comenzar a trabajar con ellos 
host_acceptance_rate                             
host_response_time                               
host_response_rate                              

```python
data1["host_acceptance_rate"] = data1["host_acceptance_rate"].str.replace("%", "", regex=False).astype(float)
```

```python
data1['host_acceptance_rate'] = data1['host_acceptance_rate'].fillna(data1['host_acceptance_rate'].mean())
data1['host_acceptance_rate'].isnull().sum()
```

```python
data1["host_response_time"] = data1["host_response_time"].fillna("unknown")
data1["host_response_time"].isnull().sum()
```

```python
data1['host_response_rate'] = data1['host_response_rate'].str.replace("%", "", regex=False).astype(float)
```

```python
data1['host_response_rate'] = data1['host_response_rate'].fillna(data1['host_response_rate'].mean())
data1['host_response_rate'].isnull().sum()
```
