## Limpieza de gastos y costos 

```python
#Importamos las librerias pandas, numpy y matplotlib respectivamente
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Carga desde un archivo .xlsx sin indice
data = pd.read_excel('Gastos y costos 20-23.xlsx',skiprows=5)
data
```

### vemos cuantos nulos tenemos 

```python
data.isnull().sum()
```

### los vemos en porcentaje 

```python
data.isnull().mean()*100
```

### copia del df 

```python
#Realizamos una copia del dataframe
data1=data.copy()
```

```python
# poliza 
#En el caso de la poliza no nos podemos inventar una por eso se registra como no dato 
#TAMBIEN POR QUE TENEMOS EL 99% DE NULOS NOS LLEVARIAMOS TODA LA BASE 
data1["POLIZA"] =data1["POLIZA"].fillna("no dato") 
```

```python
data1.isnull().sum()
```

```python
#EN GASTO LO SUSTITUIMOS POR COMPRA YA QUE HACE LOGICA CON LO QUE SE ESTA REGISTRANDO
data1["GASTO"] =data1["GASTO"].fillna("COMPRA")
```

```python
data1.isnull().sum()
```

```python
#SUSTITUIR TC POR 1 POR QUE ES EL DATO MAS REPETIDO 
data1["TC"] =data1["TC"].fillna(1)
```

```python
data1.isnull().sum()
```

```python
##folio 

data1["FOLIO"] =data1["FOLIO"].fillna(method="ffill") 
```

```python
data1.isnull().sum()
```

```python
## IVA suponemos que es de mexico entonces lo multiplico por 16 

data1['IVA'] = data1['IVA'].fillna(data1['IMPORTE'] * 0.16)
```

```python
data1.isnull().sum()
```

```python
### importe iva y tipo
### los vamos a eliminar ya que es menos de el 10% de mi base 
data2=data1.dropna()

```

```python
data2.isnull().sum()
```
