## Ventas Totales Limpio

```python
#Importamos las librerias pandas, numpy y matplotlib respectivamente
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Carga desde un archivo .csv sin indice
data = pd.read_csv('Ventas_totales.csv')
data
```

```python
data.isnull().sum()
```

```python
data.isnull().mean()*100
```

### Hacemos una copia 

```python
data1 = data.copy()
```

```python

####implementamos bfill para subtotal_ventas_alimentos_bebidas
data1["subtotal_ventas_alimentos_bebidas"] =data1["subtotal_ventas_alimentos_bebidas"].fillna(method="ffill") 

```

```python
##implementamos ffill para salon_ventas
data1["salon_ventas"] =data1["salon_ventas"].fillna(method="bfill")
```

```python
##implementamos ffill para otros
data1["otros"] =data1["otros"].fillna(method="bfill")
```

```python
##implementamos ffill para tarjetas_credito
data1["tarjetas_credito"] =data1["tarjetas_credito"].fillna(method="ffill")
```

```python
##implementamos mean para electronicos_articulos_hogar
data1["electronicos_articulos_hogar"]=data1["electronicos_articulos_hogar"].fillna(round(data1["electronicos_articulos_hogar"].mean(),1))
```

```python
data1.isnull().sum()
```

### por mean indumentaria_calzado_textiles_hogar 

```python
data1["indumentaria_calzado_textiles_hogar"]=data1["indumentaria_calzado_textiles_hogar"].fillna(round(data1["indumentaria_calzado_textiles_hogar"].mean(),1))
```

### eliminar nulos con dropna 

```python
#Eliminamos filas con nulos
data2= data1.dropna()
data2.isnull().sum()
```

```python
data2.to_csv('Ventas_totales_limpio.csv', index=False)
```
